 Teste de independência (qui-quadrado)
Pergunta “Pertencer a um cluster influencia a chance de ter lesão?”


Leitura Se p < 0.05, rejeitamos H₀: a distribuição de lesões não é igual em todos os clusters; vale investigar quais pares diferem (resíduos ajustados).

```python
import scipy.stats as st

cont = pd.crosstab(merged['cluster'], merged['injury_events'].notna())
chi2, p, *_ = st.chi2_contingency(cont)
print(f'χ² = {chi2:.2f}, p-value = {p:.4f}')
```

Modelo preditivo simples (logit)

Próximo passo Ordene os atletas pelo score predicted e entregue ao departamento médico/treinador como lista de monitoramento prioritário.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

cols_num = merged.select_dtypes('number').columns.difference(
    ['severity_idx', 'Days Missed', 'injury_events'])
X = merged[cols_num].fillna(0)
y = merged['injury_events'].notna().astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=.25,
                                                    stratify=y,
                                                    random_state=42)

clf = LogisticRegression(max_iter=1000).fit(X_train, y_train)
print('AUC:', roc_auc_score(y_test, clf.predict_proba(X_test)[:, 1]))
```
